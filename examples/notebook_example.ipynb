# mep notebook example

This is a Jupyter notebook opened in **mep**. Cells are shown as editable text in the
*percent* format (`# %%` marks a code cell, `# %% [markdown]` a markdown cell), with
each code cell's outputs drawn underneath it.

Keys (Normal mode): `Enter` runs the cell under the cursor, `Shift+Enter` runs it and
moves to the next cell, `<leader>j` opens the notebook menu (run all, insert above/below,
delete, markdown/code, move, clear, interrupt, restart kernel), `]j` / `[j` step between cells.

```python
message = "hello from the mep kernel"
print(message)
len(message)
```

## State persists across cells

The kernel is one `python3` process per notebook, so names defined above are visible below.

```python
import math
values = [math.sqrt(n) for n in range(1, 6)]
values
```

```python
import numpy as np
m = np.arange(12).reshape(3, 4)
print("shape:", m.shape)
m @ m.T
```

## Figures

`matplotlib` figures are captured inline at the end of the cell (or on `plt.show()`),
like Jupyter's inline backend. The cell below is skipped gracefully if matplotlib isn't installed.

```python
try:
    import matplotlib.pyplot as plt
    import numpy as np
except ImportError:
    print("matplotlib or numpy is not installed -- pip/nix install it to see the plot")
else:
    x = np.linspace(0, 2 * np.pi, 200)
    fig, ax = plt.subplots(figsize=(6, 3))
    ax.plot(x, np.sin(x), label="sin")
    ax.plot(x, np.cos(x), label="cos")
    ax.legend()
    ax.set_title("Inline figure from the mep kernel")
    plt.show()
```

## Errors

A raised exception becomes an error output with the traceback trimmed to the cell's own frames.

```python
def divide(a, b):
    return a / b

divide(1, 0)
```
